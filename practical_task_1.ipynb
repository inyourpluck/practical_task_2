```python
class Chocolate:
    def __init__(self, price, additive = None):
        if price < 0:
            raise ValueError('Цена не может быть меньше 0')
        else:
            self.price = price
            self.additive = additive
    def show_my_chocolate(self):
        if self.additive is None:
            print('Обычный шоколадный батончик')
        else:
            print(f'Шоколадный батончик, добавка: {self.additive}')
    def __str__(self):
        return f'Шоколадный батончик, {self.price} рублей'


mars = Chocolate(50)
print(mars)
mars.show_my_chocolate()

snikers = Chocolate(67, 'penaut')
print(snikers)
snikers.show_my_chocolate()

bounty = Chocolate(35, 'coconut')
print(bounty)
bounty.show_my_chocolate()

error_choc = Chocolate(-10)
print(error_choc)
error_choc.show_my_chocolate()
```
